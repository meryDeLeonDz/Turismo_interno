```python
from pathlib import Path
import pandas as pd
import numpy as np

# ============================================================
# 1. CARGAR DATOS
# ============================================================
# El fichero fue descargado de la web
# https://datos.canarias.es/catalogos/general/dataset/9f4355a2-d086-4384-ba72-d8c99aa2d544/resource/8ff8cc43-c00b-4513-8f42-a5b961c579e1/download/establecimientos-extrahoteleros-de-tipologia-vivienda-vacacional-inscritos-en-el-registro-genera.csv"

archivo = "/content/drive/MyDrive/Visualización/establecimientos-extrahoteleros-de-tipologia-vivienda-vacacional-inscritos-en-el-registro-genera.csv"

df = pd.read_csv(
    archivo,
    sep=";",
    encoding="utf-8"
)

print("Dimensiones:", df.shape)
print(df.head())
```

```python
# ============================================================
# 2. SELECCIONAR VARIABLES QUE NOS INTERESAN
# ============================================================

columnas = [
    "establecimiento_id",
    "establecimiento_modalidad",
    "establecimiento_tipologia",
    "establecimiento_clasificacion",
    "direccion_isla_nombre",
    "direccion_municipio_nombre",
    "plazas"
]

df_vv = df[columnas].copy()

print("\nDimensiones después de seleccionar columnas:", df_vv.shape)
display(df_vv.head())

# ============================================================
# 3. LIMPIEZA DE PLAZAS
# ============================================================

# Convertir plazas a número.
# Los valores no válidos se convierten en NaN.
df_vv["plazas"] = pd.to_numeric(df_vv["plazas"], errors="coerce")


# ============================================================
# 4. CREAR GRUPOS DE PLAZAS
# ============================================================


def agrupar_plazas(plazas):
    if pd.isna(plazas) or plazas == 0:
        return "Sin información"
    elif plazas <= 2:
        return "1-2 plazas"
    elif plazas <= 4:
        return "3-4 plazas"
    elif plazas <= 6:
        return "5-6 plazas"
    elif plazas <= 8:
        return "7-8 plazas"
    else:
        return "9 o más plazas"


df_vv["grupo_plazas"] = df_vv["plazas"].apply(agrupar_plazas)


# ============================================================
# 5. ORDENAR LA VARIABLE CATEGÓRICA
# ============================================================

orden_grupos = [
    "1-2 plazas",
    "3-4 plazas",
    "5-6 plazas",
    "7-8 plazas",
    "9 o más plazas",
    "Sin información"
]

df_vv["grupo_plazas"] = pd.Categorical(
    df_vv["grupo_plazas"],
    categories=orden_grupos,
    ordered=True
)


# ============================================================
# 6. COMPROBAR RESULTADO
# ============================================================

print("Número de registros:", len(df_vv))

print("\nDistribución por grupos de plazas:")
print(
    df_vv["grupo_plazas"]
    .value_counts()
)

print("\nValores sin información:")
print(
    df_vv[df_vv["grupo_plazas"] == "Sin información"].shape[0]
)

display(df_vv.head(10))

```

```python
# ============================================================
#  7. GUARDAR CSV PROCESADO
# ============================================================

fichero_final = "/content/viviendas_vacacionales_clean.csv"

df_vv.to_csv(
    fichero_final,
    index=False,
    encoding="utf-8-sig"
)

print("\nArchivo generado correctamente.")
print(fichero_final)
```
